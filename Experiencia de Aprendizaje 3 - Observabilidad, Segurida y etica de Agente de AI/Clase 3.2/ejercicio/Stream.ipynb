```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# Initialize the model
llm = ChatOpenAI(
    model="gpt-4o",
    api_key="YOUR_OPENAI_API_KEY_HERE",  # or set OPENAI_API_KEY env var
)

# Stream a response
for chunk in llm.stream([HumanMessage(content="Tell me a short story about a robot. Only 10 lines")]):
    print(chunk.content, end="", flush=True)

print()
```

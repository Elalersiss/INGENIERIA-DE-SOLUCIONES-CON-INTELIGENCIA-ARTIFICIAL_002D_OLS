```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

# Initialize the model
llm = ChatOpenAI(
    model="gpt-4o",
    api_key="#####################"
)

# Stream a response
for chunk in llm.stream([HumanMessage(content="Tell me a short story about a robot. Only 20 lines. Hazlo simple, rapido")]):
    print(chunk.content, end="", flush=True)

print()
```
